```python
bill_amount = float(input("Bill amount (€): "))
tip_percentage = float(input("Tip percentage: "))

tip = bill_amount * tip_percentage / 100
total = bill_amount + tip

print(f"Tip: {tip:.2f} €")
print(f"Total: {total:.2f} €")
```

 Prompt the user to enter the tip percentage and convert that input into a decimal number (float).  
  
Calculate the tip amount by multiplying the bill amount by the tip percentage divided by 100.
   
Calculate the total bill by adding the original bill amount to the calculated tip amount.  

Print the tip amount formatted to exactly two decimal places, followed by the euro symbol ("€").
